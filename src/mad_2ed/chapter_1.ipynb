### Questão 2
As escalas são conjuntos de elementos, constituído a partir de axiomas, que se pode usar para mensurar a variável de indivíduos, populações ou amostras. Os tipos de escalas são listados abaixo:

1. Escala nominal: Conjunto de símbolos que não apresentam qualquer relação entre si. Utilizada para mensurar variáveis qualitativas;
2. Escala ordinal: Conjunto de símbolos que apresentam relações de ordem, podendo ser mapeados a partir de inteiros. Utilizada para mensurar variáveis quanlitativas;
3. Escala intervalar: Conjunto de números reais separados por distâncias fixas como {0.5, 1, 1.5, ...}. O seu 0 é arbitrário e não exprime falta de informação. Utilizada para mensurar variáveis quantitativas;
4. Escala de razão: O mesmo da escala intervalar, porém seu zero exprime falta de informação e é único;

### Questão 5
Como a variável qualitativa deve ser binária, o pesquisador só conseguirá medir 2 estados. Por isso, no intuito de evitar vieses extremos, as categorias seriam: {"aceitável","não suficiente"}.

A lógica dessa escolha se dá pela sensibilidade do modelo, não há vantagem em classificar como sedentárias as pessoas que fazem exercícios de forma apenas aceitável. Classificações extremistas (saudável vs sedentário) podem acarretar em vieses por má classificação, preconceito do avaliador ou reduzir demais a amostra.
